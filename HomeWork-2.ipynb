## Part 1, Working with Numpy
    Your goal is to calculate the final grade for students in a class. 
      1. We have 10 Students in class, and 5 assessment score ( Homework, Project, Exam1, Exam2, Final Exam)for each
      student. Create a matrix containing grades of students, (rows are students, columns are assessments). Use for loop and input  function for entering the student name and grades. Enter arbitary grades from 0 to 100 for each assessment 
        2. Create a weigth vector containing weight of each assessment. Example (15%, 15%, 20%, 20%, 30% )
        3. Calculate the final grade for each students. Use Numpy to create your vector, matrix and calculating final grade.
        4. Convert final grades to letter format(90%-100% = A, 80%-89% = B, 70%-79% =C, 60%-69%=D and F for less than 60% )
        5. print out Student names and their grades

### Homework Part 1 Due: 2/18

```python
import numpy as np

# Initialize the data storage
students = []
grades_list = []
num_of_students = 10
assessments = ["Homework", "Project", "Exam1", "Exam2", "Final Exam"]

for i in range(num_of_students):
    name = input(f"Enter name of student {i+1}: ")
    students.append(name)

    current_grades = []
    for test in assessments:
        score = float(input(f" Enter score for {test}: "))
        current_grades.append(score)
    grades_list.append(current_grades)
    
# Create numpy matrix
grades_matrix = np.array(grades_list)

# Create weigh vector for percentages
weights = np.array([0.15, 0.15, 0.20, 0.20, 0.30])

# Calculate final grades using dot product
final_grades = np.dot(grades_matrix, weights)

# Convert to letters and print
print("\n --- Final Student Grades ---")
for i in range(num_of_students):
    name = students[i]
    score = final_grades[i]

    if score >= 90: letter  = "A"
    elif score >= 80: letter = "B"
    elif score >= 70: letter = "C"
    elif score >= 60: letter = "D"
    else: letter = "F"

    print(f"{name:<12}| Score:{score:>6.2f} | Grade: {letter}")
```
