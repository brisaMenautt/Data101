## Part 1

```python
import pandas as pd

#Loading and inspecting data
df = pd.read_csv('titanic.csv')

print(df.head())
print( '----------------------------------------------------------------------')

print(df.info())
print( '----------------------------------------------------------------------')
#Print variables
print(df.columns.tolist())
```

## Part 2

```python
#Fill missing Age values
df['Age'] = df['Age'].fillna(df.groupby(['Sex', 'Pclass'])['Age'].transform('median'))
print(df['Age'].head())
print('------------------------------------')
#Fill missing Embarked values
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
print(df['Embarked'].head())
```

## Part 3

```python
# 1. Create the FamilySize variable
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

# Find max, min, and mean of FamilySize variable
max_FamilySize = df['FamilySize'].max()
min_FamilySize = df['FamilySize'].min()
mean_FamilySize = df['FamilySize'].mean()
print(f'Max Family Size: {max_FamilySize}')
print(f'Min Family Size: {min_FamilySize}')
print(f'Average Family Size: {mean_FamilySize}')
print('----------------------------')

#Print Number of passengers traveling alone
alone_passengers = (df['FamilySize'] == 1).sum()
print(f"Number of passengers traveling alone: {alone_passengers}")
```

## Part 4

```python
# Data frame for females
df_female = df[df['Sex'] == 'female']

# Data frame for males
df_male = df[df['Sex'] == 'male']

# Data frame for children under 16 years old
df_under16 = df[df['Age'] < 16]

#Printing 3 data frames just made
print(df_female)
print('------------------------------------------------------------')
print(df_male)
print('------------------------------------------------------------')
print(df_under16)
print('------------------------------------------------------------')

#Calculating Survival rate for each data frame
survival_rate_female = df_female['Survived'].mean()
survival_rate_male = df_male['Survived'].mean()
survival_rate_under16 = df_under16['Survived'].mean()

#Print calculations
print(f"Female Survival Rate: {survival_rate_female}")
print(f"Male Survival Rate: {survival_rate_male}")
print(f"Under 16 Survival Rate: {survival_rate_under16}")

#Computing Survival Rate by Sex and Class
survival_by_group = df.groupby(['Sex', 'Pclass'])['Survived'].mean().unstack()
print(survival_by_group)
```

### Markdown table for sex and pclass survival rate

| Sex | First Class | Second Class | Third Class |
| :--- | :--- | :--- | :--- |
| **female** | 96.8% | 92.1% | 50% |
| **male** | 36.8% | 15.7% | 13.5% |

### Interpretation: This data shows that the female passengers had a higher rate of survival on the Titanic. It also shows that regardless of sex, the passengers with a higher-class had a higher survival rate compared to the lower-class passengers. This is most likely due to first-class passengers having greater priority on the ship.

## Part 5

```python
#Computing the average fare for each Pclass
df['FarePerPerson'] = df['Fare'] / df['FamilySize']

#Find average FarePerPerson for each class
avg_fare_per_class = df.groupby('Pclass')['FarePerPerson'].mean()

#Print results
print(avg_fare_per_class)
```

### This data shows that the first-class fares were more expensive than the second and third-class fares.
